```python
#To check the code execution time.
start = pd.Timestamp.now()
for i in range(1000):
    pass
end = pd.Timestamp.now()
print("The time taken to execute the code is : ", end-start)
```

```python
#Another method to find code execution time is >> Using time module.
import time
```

```python
start = time.time()
for i in range(1000):
    pass
end = time.time()
print("Time taken for the execution of the code is : ", end-start)
```

```python
#Another method using perf_counter
start = time.perf_counter()
for i in range(1000):
    pass
end = time.perf_counter()
print("Time taken for the execution of the code is : ", end-start)
```

```python
#To know thw date and time before a week exactly.
start = pd.Timestamp.now() - pd.DateOffset(weeks=1)
start
```

```python
#To know the date and time exactly after a week.
end = pd.Timestamp.now() + pd.DateOffset(weeks=1)
end
```

```python
#Timedelta means time difference
end - start 
```
