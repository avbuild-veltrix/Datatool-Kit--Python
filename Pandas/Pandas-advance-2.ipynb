```python
# How many passenger survived or not survived for sex and pclass.
df.groupby(["Sex","Pclass"])["Survived"].sum()
```

```python
# How many passenger survived or not survived for sex and pclass and convert it into dataframe.
df.groupby(["Sex","Pclass"])["Survived"].sum().to_frame()
```

```python
# How many passenger survived or not survived for sex and pclass and convert it into dataframe.
df.groupby(["Sex","Pclass"])["Survived"].sum().to_frame().reset_index()
```

```python
# How many passenger survived or not survived for sex and pclass and convert it into dataframe.
df.groupby(["Sex","Pclass"])["Survived"].sum().to_frame().unstack()
```
