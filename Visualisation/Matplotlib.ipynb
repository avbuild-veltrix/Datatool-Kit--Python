```python
# Pie chart >> Use to understand percentage distribution of data.
langs = ["Python", "C","C++"]
students = [100, 20, 40]
plt.pie(students, labels = langs, autopct = "%1.1f%%")
plt.title("Pie Chart")
plt.show()
```

```python
# If you want o focus on one segment then
explode = (0.0, 0.1, 0.1)
langs = ["Python", "C","C++"]
students = [100, 20, 40]
plt.pie(students, labels = langs, autopct = "%1.1f%%", explode = explode)
plt.title("Pie Chart")
plt.show()
```

```python
# If you want o focus on one segment then
langs = ["Python", "C","C++"]
students = [100, 20, 40]
plt.pie(students, labels = langs, autopct = "%1.1f%%", explode = (0.005, 0.1, 0.05))
plt.title("Pie Chart")
plt.show()
```

```python
# If you want o focus on one segment then
langs = ["Python", "C","C++"]
students = [100, 20, 40]
plt.pie(students, labels = langs, autopct = "%1.1f%%", explode = (0.005, 0.1, 0.1), shadow = True)
plt.title("Pie Chart")
plt.show()
```

```python
# Area Plot >> Showshow some data changng over time. >> Also called as stacked area plot.

years = [2021, 2018, 2022, 2013, 2017]
cat1 = [20,25,30,28,35]
cat2 = [10,15,20,18,25]
cat3 = [5 , 8, 10, 9, 15]

plt.stackplot(years, cat1, cat2, cat3, labels = ["Category1" , "Category2", "Category3"], colors = ['skyblue' , 'lightgreen', 'red'])
plt.title("Stacked Area Plot")
plt.xlabel("Values")
plt.ylabel("Years")
plt.legend()
plt.show()
```
